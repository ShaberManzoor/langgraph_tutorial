```python
from langgraph.graph import StateGraph, START, END
from typing import TypedDict
```

```python
# making a state
class BatsmanState(TypedDict):
    runs: int
    balls: int
    fours: int
    six: int
    
    sr: float
    bpb: float
    boundryPercent: float
    summary: str
```

```python
def calculate_sr(state: BatsmanState) -> BatsmanState :
    balls = state['balls']
    runs = state['runs']
    
    sr = runs / balls * 100
    state['sr'] = sr
    
    return {'sr': sr}
```

```python
def cal_bpb(state: BatsmanState) -> BatsmanState: 
    balls = state['balls']
    fours = state['fours']
    sixes = state['six']
    
    boundry = fours + sixes
    state['bpb'] = balls/boundry
    
    return {'bpb': state['bpb']}
```

```python
def cal_boundryPer(state: BatsmanState) -> BatsmanState: 
    runs = state['runs']
    fours = state['fours']
    sixes = state['six']
    
    state['boundryPercent'] = ((fours*4 + sixes*6)/runs) * 100
    
    return {'boundryPercent': state['boundryPercent']}
```

```python
def summary(state: BatsmanState)-> BatsmanState: 
    
    summary = f"""
    Strike Rate - {state['sr']} \n
    Balls per Boundry - {state['bpb']} \n
    Boundry Percent - {state['boundryPercent']}
    """
    
    state['summary'] = summary
    return state
```

```python
# Building a Graph

graph = StateGraph(BatsmanState)

graph.add_node('calculate_sr', calculate_sr)
graph.add_node('cal_bpb', cal_bpb)
graph.add_node('cal_boundryPer', cal_boundryPer)
graph.add_node('summary', summary)

graph.add_edge(START, 'calculate_sr')
graph.add_edge(START, 'cal_bpb')
graph.add_edge(START, 'cal_boundryPer')

graph.add_edge('calculate_sr', 'summary')
graph.add_edge('cal_bpb', 'summary')
graph.add_edge('cal_boundryPer', 'summary')

graph.add_edge('summary', END)

workflow = graph.compile()
```

```python
initial_state = {
    'runs': 100,
    'balls': 40,
    'fours': 8,
    'six': 11
}

workflow.invoke(initial_state)
```
